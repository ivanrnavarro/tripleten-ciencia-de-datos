```python
# Preparación inicial de los datos antes de iniciar entrenamiento de modelos

features_0 = geo_data_0.drop(['id', 'product'], axis=1)
target_0 = geo_data_0['product']

features_1 = geo_data_1.drop(['id', 'product'], axis=1)
target_1 = geo_data_1['product']

features_2 = geo_data_2.drop(['id', 'product'], axis=1)
target_2 = geo_data_2['product']
```

## 2. Entrenamiento y prueba del modelo de regresión lineal



```python
# Definición del random_state
state = 792604

# Definición de la función para entrenamiento del modelo de regresión lineal en cualquier conjunto de datos:
def linear_regression_model(features, target, state):
    # Formación conjuntos de entrenamiento y validación
    features_train, features_valid, target_train, target_valid = train_test_split(features, target, test_size=0.25, random_state=state)

    # Entrenamiento del modelo de Regresión Lineal
    model = LinearRegression()
    model.fit(features_train, target_train)

    # Predicciones para el grupo de validación
    predictions_valid = model.predict(features_valid)

    # Volumen medio predicho
    product_valid = predictions_valid.mean()

    # calculo RSME
    rmse = mean_squared_error(target_valid, predictions_valid)**0.5
    
    return predictions_valid, target_valid, product_valid, rmse

# Aplicación del modelo a geo_data_0
predictions_valid_0, target_valid_0, product_valid_0, rmse_0 = linear_regression_model(features_0, target_0, state)

# Datos finales
print('geo_data_0:')
print('Volumen promedio de reservas del modelo: ', product_valid_0)
print('RMSE: ', rmse_0)
print()

# Aplicación del modelo a geo_data_1
predictions_valid_1, target_valid_1, product_valid_1, rmse_1 = linear_regression_model(features_1, target_1, state)

# Datos finales
print('geo_data_1:')
print('Volumen promedio de reservas del modelo: ', product_valid_1)
print('RMSE: ', rmse_1)
print()

# Aplicación del modelo a geo_data_2
predictions_valid_2, target_valid_2, product_valid_2, rmse_2 = linear_regression_model(features_2, target_2, state)

# Datos finales
print('geo_data_2:')
print('Volumen promedio de reservas del modelo: ', product_valid_2)
print('RMSE: ', rmse_2)
```

##### Observaciones:

Se ejecuta el entrenamiento del modelo de regresión lineal para todos los conjuntos de datos. cada conjunto fue particionado en juegos de entrenamiento y validación en proporción 0.75:0.25. 

Para geo_data_0:
- Se obtiene un volumen promedio de reservas de 92.457 miles de barriles de petroleo, que al compararse con los datos iniciales del juego de datos (92.5 miles de barriles) muestra una similitud, indicando que el modelo opera bien.
- No obstante, se obtiene un RMSE de 37.8, lo cual es un valor alto, que sin embargo es menor que la desviación estandar encontrada en el juego de datos original (44.3). Esto nos da a entender que si bien el modelo aun necesita algunos ajustes adicionales, esta operando con relativa confiabilidad.

Para geo_data_1:
- El volumen promedio predicho (68.293 miles de barriles) y real (68.8 miles de barriles), siguen guardando la misma cercanía.
- El RMSE fue de 0.89, la desviación estándar de los datos es de 45.9. Esta diferencia se convierte en una señal de alerta sobre el modelo ya que hay un cambio importante de una región a otra.
  
Para Geo_data_2:

- El RMSE de 39.9 y la desviación estándar de los datos de 44.8 sigue indicando ruido que aun se requiere mejorar.

En términos generales, el modelo guarda buena respuesta en predicción media de capacidad de producción, sin embargo, en valores puntuales no opera con relativa confiabilidad. Se debe tener en cuenta que la variabilidad de un pozo a otro en la vida real es alta y esto finalmente lo toma el modelo.

Aparentemente el modelo opera mejor con los datos del segundo grupo, pero tengamos en cuenta que desde la observación inicial de los juegos de datos se encontró amplia variación de estos valores.

## 4. Cálculo de ganancias de grupo de pozos explorados

```python
def selection(target, predictions, count, revenue, invest):
    # Conversión del archivo de predicciones a serie de Pandas
    predictions = pd.Series(predictions, index=target.index)
    # Organizar 200 pozos de mayor capacidad en orden descendente
    predictions_sorted = predictions.sort_values(ascending=False)
    # Seleccionar los datos de los pozos con mejores posibilidades
    selected = target[predictions_sorted.index[:count]]
    # Producción esperada de los mejores 200 pozos
    productions = selected.sum()
    # Ingresos esperados por los pozos explorados
    income = productions * revenue
    # Ganancias esperadas
    earnings = income - invest
    return selected, productions, income, earnings

# Aplicación de la función para geo_data_0
selected_0, productions_0, income_0, earnings_0 = selection(target_valid_0, predictions_valid_0, 200, revenue, invest)
print('geo_data_0')
print('Volumen objetivo de reservas de los mejores 200 pozos (miles de barriles de petroleo): ', productions_0)
print()
print('Ingresos esperados por los mejores 200 pozos (USD millions): ', income_0/1000000)
print()
print('Ganacias esperadas (USD millions): ', earnings_0/1000000)
print()

# Aplicación de la función para geo_data_1
selected_1, productions_1, income_1, earnings_1 = selection(target_valid_1, predictions_valid_1, 200, revenue, invest)
print('geo_data_1')
print('Volumen objetivo de reservas de los mejores 200 pozos (miles de barriles de petroleo): ', productions_1)
print()
print('Ingresos esperados por los mejores 200 pozos (USD millions): ', income_1/1000000)
print()
print('Ganacias esperadas (USD millions): ', earnings_1/1000000)
print()

# Aplicación de la función para geo_data_2
selected_2, productions_2, income_2, earnings_2 = selection(target_valid_2, predictions_valid_2, 200, revenue, invest)
print('geo_data_2')
print('Volumen objetivo de reservas de los mejores 200 pozos (miles de barriles de petroleo): ', productions_2)
print()
print('Ingresos esperados por los mejores 200 pozos (USD millions): ', income_2/1000000)
print()
print('Ganacias esperadas (USD millions): ', earnings_2/1000000)
print()
```

#### Observaciones

Se ha realizado el calculo de ganancia para los datos predichos de cada región disponible:

- Para la región 0 las ganancias son de 31.8 millones de dolares
- Para la región 1 las ganancias son de 24.1 millones de dolares
- Para la región 2 las ganancias son de 25 millones de dolares

Basados en estas ganancias, la región geo_data_0 debe ser la seleccionada para realizar la exploración.

Mediante técnicas de bootstrapping se procederá a continuación a corroborar esta elección, teniendo en cuenta la dispersión de los datos que se tienen para todo el juego de datos y especialmente por el RmsE que presenta el modelo para dicha región.

## Cálculo de riesgos y ganancias para cada región

```python
# Definición tamaño de muestra e iteraciones a realizar
sub_sample = 500
iterac = 1000

# Función para el cálculo de ganancias promedio, intérvalo de confianza y riesgo de pérdidas
def revenues(target, predictions):
    # Conversión del archivo de predicciones a serie de Pandas
    predictions = pd.Series(predictions, index=target.index)
    # Bootstrapping
    earnings = []
    for i in range(iterac):
        # Formación submuestras
        target_subsample = target.sample(n=sub_sample, replace=True, random_state=i)
        predictions_subsample = predictions[target_subsample.index]
        # Cálculo de ganancias
        earnings.append(selection(target_subsample, predictions_subsample, 200, revenue, invest)[3])
    # Conversion de earnings a serie de pandas
    earnings = pd.Series(earnings)
    # Ingresos promedio
    earnings_mean = earnings.mean()
    # Intervalo de confianza
    confidence_interval = st.t.interval(confidence=0.95, df=len(target_subsample)-1, loc=earnings_mean, scale=earnings.sem())
    # Cálculo de riesgo de pérdidas
    risk = (earnings < 0).mean()
    return (earnings_mean, confidence_interval, risk)

# aplicación en geo_data_0
earn_mean_0, interval_0, risk_0 = revenues(target_valid_0, predictions_valid_0)
print('geo_data_0')
print('El promedio de ganancias es (USD Millions): ', earn_mean_0/1000000)
print('El porcentaje de pérdidas es: ', risk_0*100)
print('El intérvalo de confianza es ', interval_0[0]/1000000, '-', interval_0[1]/1000000 )
print()

# aplicación en geo_data_1
earn_mean_1, interval_1, risk_1 = revenues(target_valid_1, predictions_valid_1)
print('geo_data_1')
print('El promedio de ganancias es (USD Millions): ', earn_mean_1/1000000)
print('El porcentaje de pérdidas es: ', risk_1*100)
print('El intérvalo de confianza es ', interval_1[0]/1000000, '-', interval_1[1]/1000000)
print()

# aplicación en geo_data_2
earn_mean_2, interval_2, risk_2 = revenues(target_valid_2, predictions_valid_2)
print('geo_data_2')
print('El promedio de ganancias es (USD Millions): ', earn_mean_2/1000000)
print('El porcentaje de pérdidas es: ', risk_2*100)
print('El intérvalo de confianza es ', interval_2[0]/1000000, '-', interval_2[1]/1000000)
```

#### Observaciones
- Realizando la técnica de bootstrapping a los juegos de validación, seleccionando 500 pozos aleatoriamente y escogiendo los mejores 200 los valores cambian con respecto al punto anterior.
- Para la región 0, el promedio de ganancias sería de 5.7 millones de dolares, y hay un riesgo de pérdidas de 2.6% con un intervalo de confianza de 5.5-5.9 millones de dolares.
- Para la región 1, el promedio de ganancias es de 6.3 millones de dolares y el riesgo de pérdidas es de 0.2% con un intérvalo de confianza de 6.1-6.5 millones de dolares.
- En la región 2, la gananacia promedio es de 5.5 millones de dolares y el riesgo de pérdida es de 3.5% con un intérvalo de confianza de 5.3-5.8 millones de dolares.
